```python
rfm = cleaned_df[
    [
        "ID",
        "Recency",
        "Total_Purchases",
        "Total_Spending"
    ]
].copy()

rfm.columns = [
    "Customer_ID",
    "Recency",
    "Frequency",
    "Monetary"
]

rfm.head()
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm.head()
```

```python
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm.head()
```

```python
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm.head()
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

rfm.head()
```

```python
def segment_customer(row):

    if (
        row["R_Score"] >= 4
        and row["F_Score"] >= 4
        and row["M_Score"] >= 4
    ):
        return "Champions"

    elif (
        row["R_Score"] >= 4
        and row["F_Score"] >= 3
    ):
        return "Loyal Customers"

    elif (
        row["R_Score"] >= 4
        and row["M_Score"] >= 3
    ):
        return "Potential Loyalists"

    elif row["R_Score"] >= 4:
        return "New Customers"

    elif (
        row["R_Score"] <= 2
        and row["F_Score"] >= 3
        and row["M_Score"] >= 3
    ):
        return "At Risk"

    else:
        return "Needs Attention"
```

```python
rfm["Customer_Segment"] = rfm.apply(
    segment_customer,
    axis=1
)

rfm.head()
```

```python
segment_count = rfm[
    "Customer_Segment"
].value_counts()

print(segment_count)
```

```python
cleaned_df = cleaned_df.merge(
    rfm[
        [
            "Customer_ID",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score",
            "Customer_Segment"
        ]
    ],
    left_on="ID",
    right_on="Customer_ID",
    how="left"
)

cleaned_df.drop(
    columns=["Customer_ID"],
    inplace=True
)

cleaned_df.head()
```
